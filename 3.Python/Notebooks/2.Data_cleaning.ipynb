```python
import pandas as pd
import numpy as np
import os

data_path = "../../2. Data/Raw"

country = pd.read_csv(os.path.join(data_path, "dim_country.csv"))
date = pd.read_csv(os.path.join(data_path, "dim_date.csv"))
industry = pd.read_csv(os.path.join(data_path, "dim_industry.csv"))
skill = pd.read_csv(os.path.join(data_path, "dim_skill_category.csv"))
fact = pd.read_csv(os.path.join(data_path, "fact_workforce_ai_index.csv"))
```

```python
fact["avg_wage_change_pct"] = fact["avg_wage_change_pct"].fillna(
    fact["avg_wage_change_pct"].mean()
)

fact["ai_tool_usage_hours_per_week"] = fact["ai_tool_usage_hours_per_week"].fillna(
    fact["ai_tool_usage_hours_per_week"].mean()
)
```

```python
fact.isnull().sum()
```

```python
clean_path = "../../2. Data/Cleaned"

country.to_csv(os.path.join(clean_path, "dim_country_clean.csv"), index=False)

date.to_csv(os.path.join(clean_path, "dim_date_clean.csv"), index=False)

industry.to_csv(os.path.join(clean_path, "dim_industry_clean.csv"), index=False)

skill.to_csv(os.path.join(clean_path, "dim_skill_category_clean.csv"), index=False)

fact.to_csv(os.path.join(clean_path, "fact_workforce_ai_index_clean.csv"), index=False)
```
