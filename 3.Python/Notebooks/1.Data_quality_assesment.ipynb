# Data Quality Assesment 

## purpose 
- To understand the dataset 
- Check data types
- Check missing values
- Check duplicate records
- Check unique values

```python
import pandas as pd 
import numpy as np
```

```python
import os 
data_path = "../../2. Data/Raw"
```

```python
country = pd.read_csv(os.path.join(data_path, "dim_country.csv"))
date = pd.read_csv(os.path.join(data_path, "dim_date.csv"))
industry = pd.read_csv(os.path.join(data_path, "dim_industry.csv"))
skill = pd.read_csv(os.path.join(data_path, "dim_skill_category.csv"))
fact = pd.read_csv(os.path.join(data_path, "fact_workforce_ai_index.csv"))
```

```python
print("Country:", country.shape)
print("Date:", date.shape)
print("Industry:", industry.shape)
print("Skill:", skill.shape)
print("Fact:", fact.shape)
```

## dataset structure


```python
datasets = {
    "Country": country,
    "Date": date,
    "Industry": industry,
    "Skill": skill,
    "Fact": fact
}

for name, df in datasets.items():
    print("=" * 60)
    print(name.upper())
    print("=" * 60)

    display(df.head())

    print("\nShape:", df.shape)

    print("\nColumn Names:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\n")
```

## Dataset Info


```python
for name, df in datasets.items():
    print("=" * 80)
    print(f"{name.upper()} DATASET INFO")
    print("=" * 80)

    df.info()

    print("\n")
```
